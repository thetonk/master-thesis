# Feature importance

Purpose of these experiments is to find out which features do models focus more on to classify the samples of each dataset, using SHAP values. These experiments were also conducted for the case we remove features _regardless_ of the model type. Experiments configuration:

## Transformer
- Training: 30 epochs with early stopping
- Removed features:
    - Timestamp
    - Src IP
    - Dst IP
    - Idle Mean
    - Idle Min
    - Idle Max

## LSTM 
- Training: 5 epochs with early stopping
- Removed features:
    - Timestamp
    - Fwd Seg Size Min

For each dataset, 5 runs of 10-fold cross validation have taken place, to provide statistically accurate results. The following datasets were used:

- CIC-UNSW-NB15
- IDS-2018
- IoT-23
- IoT-DS2
- IoTID20
- MQTT-IoT-IDS2020
- USTC-TFC-2016

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import os

RESULTS_DIR = "../../results/reports"
LSTM_RESULTS_DIR = os.path.join(RESULTS_DIR, "lstm", "feature_importance")
TRANSFORMER_RESULTS_DIR = os.path.join(RESULTS_DIR, "transformer", "feature_importance")

dataset_name = "CIC-UNSW-NB15"
cic_unsw_nb15_lstm_path = os.path.join(LSTM_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_lstm_CIC-UNSW-NB15-binclass.csv")
cic_unsw_nb15_transformer_path = os.path.join(TRANSFORMER_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_transformer_CIC-UNSW-NB15-binclass.csv")
cic_unsw_nb15_lstm_removed_path = os.path.join(LSTM_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_lstm_CIC-UNSW-NB15-binclass_removed.csv")
cic_unsw_nb15_transformer_removed_path = os.path.join(TRANSFORMER_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_transformer_CIC-UNSW-NB15-binclass_removed.csv")
cic_unsw_nb15_lstm_df = pd.read_csv(cic_unsw_nb15_lstm_path)
cic_unsw_nb15_transformer_df = pd.read_csv(cic_unsw_nb15_transformer_path)
cic_unsw_nb15_lstm_removed_df = pd.read_csv(cic_unsw_nb15_lstm_removed_path)
cic_unsw_nb15_transformer_removed_df = pd.read_csv(cic_unsw_nb15_transformer_removed_path)

dataset_name = "IDS-2018"
ids_2018_lstm_path = os.path.join(LSTM_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_lstm_ids-2018-binclass.csv")
ids_2018_transformer_path = os.path.join(TRANSFORMER_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_transformer_ids-2018-binclass.csv")
ids_2018_lstm_removed_path = os.path.join(LSTM_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_lstm_ids-2018-binclass_removed.csv")
ids_2018_transformer_removed_path = os.path.join(TRANSFORMER_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_transformer_ids-2018-binclass_removed.csv")
ids_2018_lstm_df = pd.read_csv(ids_2018_lstm_path)
ids_2018_transformer_df = pd.read_csv(ids_2018_transformer_path)
ids_2018_lstm_removed_df = pd.read_csv(ids_2018_lstm_removed_path)
ids_2018_transformer_removed_df = pd.read_csv(ids_2018_transformer_removed_path)

dataset_name = "IoT-23"
iot_23_lstm_path = os.path.join(LSTM_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_lstm_IoT-23-binclass.csv")
iot_23_transformer_path = os.path.join(TRANSFORMER_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_transformer_IoT-23-binclass.csv")
iot_23_lstm_removed_path = os.path.join(LSTM_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_lstm_IoT-23-binclass_removed.csv")
iot_23_transformer_removed_path = os.path.join(TRANSFORMER_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_transformer_IoT-23-binclass_removed.csv")
iot_23_lstm_df = pd.read_csv(iot_23_lstm_path)
iot_23_transformer_df = pd.read_csv(iot_23_transformer_path)
iot_23_lstm_removed_df = pd.read_csv(iot_23_lstm_removed_path)
iot_23_transformer_removed_df = pd.read_csv(iot_23_transformer_removed_path)

dataset_name = "IoT-DS2"
iot_ds2_lstm_path = os.path.join(LSTM_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_lstm_IoT-DS2-binclass.csv")
iot_ds2_transformer_path = os.path.join(TRANSFORMER_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_transformer_IoT-DS2-binclass.csv")
iot_ds2_lstm_removed_path = os.path.join(LSTM_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_lstm_IoT-DS2-binclass_removed.csv")
iot_ds2_transformer_removed_path = os.path.join(TRANSFORMER_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_transformer_IoT-DS2-binclass_removed.csv")
iot_ds2_lstm_df = pd.read_csv(iot_ds2_lstm_path)
iot_ds2_transformer_df = pd.read_csv(iot_ds2_transformer_path)
iot_ds2_lstm_removed_df = pd.read_csv(iot_ds2_lstm_removed_path)
iot_ds2_transformer_removed_df = pd.read_csv(iot_ds2_transformer_removed_path)

dataset_name = "IoTID20"
iotid20_lstm_path = os.path.join(LSTM_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_lstm_iotid20-binclass.csv")
iotid20_transformer_path = os.path.join(TRANSFORMER_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_transformer_iotid20-binclass.csv")
iotid20_lstm_removed_path = os.path.join(LSTM_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_lstm_iotid20-binclass_removed.csv")
iotid20_transformer_removed_path = os.path.join(TRANSFORMER_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_transformer_iotid20-binclass_removed.csv")
iotid20_lstm_df = pd.read_csv(iotid20_lstm_path)
iotid20_transformer_df = pd.read_csv(iotid20_transformer_path)
iotid20_lstm_removed_df = pd.read_csv(iotid20_lstm_removed_path)
iotid20_transformer_removed_df = pd.read_csv(iotid20_transformer_removed_path)

dataset_name = "MQTT-IoT-IDS2020"
mqtt_iot_ids2020_lstm_path = os.path.join(LSTM_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_lstm_mqtt-iot-ids2020-merged.csv")
mqtt_iot_ids2020_transformer_path = os.path.join(TRANSFORMER_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_transformer_mqtt-iot-ids2020-merged.csv")
mqtt_iot_ids2020_lstm_removed_path = os.path.join(LSTM_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_lstm_mqtt-iot-ids2020-merged_removed.csv")
mqtt_iot_ids2020_transformer_removed_path = os.path.join(TRANSFORMER_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_transformer_mqtt-iot-ids2020-merged_removed.csv")
mqtt_iot_ids2020_lstm_df = pd.read_csv(mqtt_iot_ids2020_lstm_path)
mqtt_iot_ids2020_transformer_df = pd.read_csv(mqtt_iot_ids2020_transformer_path)
mqtt_iot_ids2020_lstm_removed_df = pd.read_csv(mqtt_iot_ids2020_lstm_removed_path)
mqtt_iot_ids2020_transformer_removed_df = pd.read_csv(mqtt_iot_ids2020_transformer_removed_path)

dataset_name = "USTC-TFC-2016"
ustc_tfc_2016_lstm_path = os.path.join(LSTM_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_lstm_usfc-tfc-2016-binclass.csv")
ustc_tfc_2016_transformer_path = os.path.join(TRANSFORMER_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_transformer_usfc-tfc-2016-binclass.csv")
ustc_tfc_2016_lstm_removed_path = os.path.join(LSTM_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_lstm_usfc-tfc-2016-binclass_removed.csv")
ustc_tfc_2016_transformer_removed_path = os.path.join(TRANSFORMER_RESULTS_DIR, "unified_removed_features", dataset_name.lower(), "attack_shap_values_transformer_usfc-tfc-2016-binclass_removed.csv")
ustc_tfc_2016_lstm_df = pd.read_csv(iot_ds2_lstm_path)
ustc_tfc_2016_transformer_df = pd.read_csv(iot_ds2_transformer_path)
ustc_tfc_2016_lstm_removed_df = pd.read_csv(iot_ds2_lstm_removed_path)
ustc_tfc_2016_transformer_removed_df = pd.read_csv(iot_ds2_transformer_removed_path)

def plot_feature_importance(df: pd.DataFrame, dataset_name: str, model_name: str, removed_features: bool = False):
    plot_title = f"Feature importance for {model_name} on {dataset_name} dataset"
    df = df.sort_values("SHAP_value", ascending=False)
    bar_plot = df.head(25).plot(x="Feature", y="SHAP_value", kind="bar", rot=90, title=plot_title, figsize=(12,5), width=0.8)
    bar_plot = bar_plot.get_figure()
    bar_plot.tight_layout()
    if removed_features:
        bar_plot.savefig(f"{dataset_name.lower()}-{model_name.lower()}-removed.pdf")
    else:
        bar_plot.savefig(f"{dataset_name.lower()}-{model_name.lower()}.pdf")
```

## Feature importance summary
Feature importance per model, across all datasets
### All network specific features included

```python
normalizer = MinMaxScaler()

features = iot_23_lstm_df["Feature"]
shap_value_list = [features, ids_2018_lstm_df["SHAP_value"], cic_unsw_nb15_lstm_df["SHAP_value"],
                   ustc_tfc_2016_lstm_df["SHAP_value"], iot_23_lstm_df["SHAP_value"], iot_ds2_lstm_df["SHAP_value"], 
                   iotid20_lstm_df["SHAP_value"], mqtt_iot_ids2020_lstm_df["SHAP_value"]]
summary_df = pd.concat(shap_value_list, axis=1, keys=["Feature", "ids-2018", "cic-unsw-nb15", "ustc-tfc-2016", "iot-23", "iot-ds2", "iotid20", "mqtt-iot-ids2020"])
summary_df["Total SHAP value"] = summary_df.sum(axis=1, numeric_only=True)
summary_normalized_df = pd.DataFrame({"SHAP value": summary_df["Total SHAP value"]})
normalized_shap_values = normalizer.fit_transform(summary_normalized_df).flatten()
summary_normalized_df = pd.DataFrame({"Features": features, "SHAP values": normalized_shap_values})
summary_normalized_df = summary_normalized_df.sort_values("SHAP values", ascending=False)
lstm_plot = summary_normalized_df.head(25).plot(x="Features", y="SHAP values", kind="bar", rot=90, title="Feature importance for LSTM", figsize=(12,5), width=0.8)
lstm_plot = lstm_plot.get_figure()
lstm_plot.tight_layout()
lstm_plot.savefig("feature-importance-summary-lstm.pdf")

features = iot_23_transformer_df["Feature"]
shap_value_list = [features, ids_2018_transformer_df["SHAP_value"], cic_unsw_nb15_transformer_df["SHAP_value"],
                   ustc_tfc_2016_transformer_df["SHAP_value"], iot_23_transformer_df["SHAP_value"], iot_ds2_transformer_df["SHAP_value"], 
                   iotid20_transformer_df["SHAP_value"], mqtt_iot_ids2020_transformer_df["SHAP_value"]]
summary_df = pd.concat(shap_value_list, axis=1, keys=["Feature", "ids-2018", "cic-unsw-nb15", "ustc-tfc-2016", "iot-23", "iot-ds2", "iotid20", "mqtt-iot-ids2020"])
summary_df["Total SHAP value"] = summary_df.sum(axis=1, numeric_only=True)
summary_normalized_df = pd.DataFrame({"SHAP value": summary_df["Total SHAP value"]})
normalized_shap_values = normalizer.fit_transform(summary_normalized_df).flatten()
summary_normalized_df = pd.DataFrame({"Features": features, "SHAP values": normalized_shap_values})
summary_normalized_df = summary_normalized_df.sort_values("SHAP values", ascending=False)
transformer_plot = summary_normalized_df.head(25).plot(x="Features", y="SHAP values", kind="bar", rot=90, title="Feature importance for Transformer", figsize=(12,5), width=0.8)
transformer_plot = transformer_plot.get_figure()
transformer_plot.tight_layout()
transformer_plot.savefig("feature-importance-summary-transformer.pdf")
```
