# Feature importance

Purpose of these experiments is to find out which features do models focus more on to classify the samples of each dataset, using SHAP values. These experiments were also conducted for the case we remove features according to the model type. Experiments configuration:

## Transformer
- Training: 30 epochs with early stopping
- Removed features:
    - Timestamp
    - Src IP
    - Dst IP
    - Idle Mean
    - Idle Min
    - Idle Max

## LSTM 
- Training: 5 epochs with early stopping
- Removed features:
    - Timestamp
    - Fwd Seg Size Min

For each dataset, 5 runs of 10-fold cross validation have taken place, to provide statistically accurate results. The following datasets were used:

- CIC-UNSW-NB15
- IDS-2018
- IoT-23
- IoT-DS2
- IoTID20
- MQTT-IoT-IDS2020
- USTC-TFC-2016

```python
import pandas as pd
import os

RESULTS_DIR = "../../results/reports"
LSTM_RESULTS_DIR = os.path.join(RESULTS_DIR, "lstm", "feature_importance")
TRANSFORMER_RESULTS_DIR = os.path.join(RESULTS_DIR, "transformer", "feature_importance")

def plot_feature_importance(csv_path: str, dataset_name: str, model_name: str, removed_features: bool = False):
    plot_title = f"Feature importance for {model_name} on {dataset_name} dataset"
    df = pd.read_csv(csv_path).sort_values("SHAP_value", ascending=False)
    bar_plot = df.head(25).plot(x="Feature", y="SHAP_value", kind="bar", rot=90, title=plot_title, figsize=(12,5), width=0.8)
    bar_plot = bar_plot.get_figure()
    bar_plot.tight_layout()
    if removed_features:
        bar_plot.savefig(f"{dataset_name.lower()}-{model_name.lower()}-removed.pdf")
    else:
        bar_plot.savefig(f"{dataset_name.lower()}-{model_name.lower()}.pdf")
```

## All network specific features included

### CIC-UNSW-NB15

```python
dataset_name = "CIC-UNSW-NB15"
lstm_df_path = os.path.join(LSTM_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_lstm_CIC-UNSW-NB15-binclass.csv")
transformer_df_path = os.path.join(TRANSFORMER_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_transformer_CIC-UNSW-NB15-binclass.csv")

plot_feature_importance(lstm_df_path, dataset_name, "LSTM", removed_features=False)
plot_feature_importance(transformer_df_path, dataset_name, "Transformer", removed_features=False)
```

```python
dataset_name = "IDS-2018"
lstm_df_path = os.path.join(LSTM_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_lstm_ids-2018-binclass.csv")
transformer_df_path = os.path.join(TRANSFORMER_RESULTS_DIR, "with_network_specific_features", dataset_name.lower(), "attack_shap_values_transformer_ids-2018-binclass.csv")

plot_feature_importance(lstm_df_path, dataset_name, "LSTM", removed_features=False)
plot_feature_importance(transformer_df_path, dataset_name, "Transformer", removed_features=False)
```
