# Dataset categories

Exploring the available cyberattack and malware categories available on each dataset.

```python
import pandas as pd
import os

def show_categories(column_name: str, csv_file: str, clipboard: bool = False):
    df = pd.read_csv(csv_file, usecols=[column_name])
    samples_per_category = df[column_name].value_counts()
    if clipboard:
        samples_per_category.to_clipboard()
    print(f"Total rows: {df.shape[0]}")
    print(samples_per_category)

DATASET_DIRECTORY = "/mnt/btrfs/datasets/cybersecurity/multiclass"
```

## After malware category standardization
### Unified

```python
dataset_file = os.path.join(DATASET_DIRECTORY, "unified", "cybersecurity-unified-multiclass.csv")
show_categories("Label", dataset_file)
```

As a result, if we want to perform balanced multiclass malware classification, we will need to have $16767$ samples for each of the $9$ availables classes, resulting in a total of $16767 \times 9 = 150903$ samples from the unified dataset.
